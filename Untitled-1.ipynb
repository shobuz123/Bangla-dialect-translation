```python
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer
import torch

# NLLB model এর path (সবচেয়ে ভালো, BLEU 91.76)
nllb_path = r"C:\Users\SAMSUNG GALAXY\Desktop\dialect_project\mate_2models (1)\model_nllb200"

print("মডেল লোড হচ্ছে... (প্রথমবার একটু সময় লাগবে)")
tok = AutoTokenizer.from_pretrained(nllb_path)
model = AutoModelForSeq2SeqLM.from_pretrained(nllb_path)
model.eval()
print("✅ NLLB লোড হয়েছে!")

# অনুবাদ ফাংশন
def translate(text, direction="dialect", dialect="sylheti"):
    if direction == "dialect":
        inp = f"<{dialect}> translate to dialect: {text}"
    else:
        inp = f"<{dialect}> translate to standard: {text}"
    ids = tok(inp, return_tensors="pt", truncation=True, max_length=64)
    with torch.no_grad():
        out = model.generate(**ids, max_length=64, num_beams=4)
    return tok.decode(out[0], skip_special_tokens=True)

# টেস্ট — standard Bangla → Sylheti
text = "তুমি কেমন আছো?"
result = translate(text, direction="dialect", dialect="sylheti")
print(f"\nইনপুট (standard): {text}")
print(f"আউটপুট (sylheti): {result}")
```

```python
# mBART লোড করো (তোমার দ্বিতীয় বেস্ট, BLEU 76.03)
mbart_path = r"C:\Users\SAMSUNG GALAXY\Desktop\dialect_project\shobuz_3models\model_mbart50"

print("mBART লোড হচ্ছে...")
mbart_tok = AutoTokenizer.from_pretrained(mbart_path)
mbart_model = AutoModelForSeq2SeqLM.from_pretrained(mbart_path)
mbart_model.eval()
print("✅ mBART লোড হয়েছে!")

# দুইটা মডেল দিয়ে অনুবাদের ফাংশন
def translate_both(text, direction="dialect", dialect="sylheti"):
    if direction == "dialect":
        inp = f"<{dialect}> translate to dialect: {text}"
    else:
        inp = f"<{dialect}> translate to standard: {text}"

    # NLLB
    ids = tok(inp, return_tensors="pt", truncation=True, max_length=64)
    with torch.no_grad():
        out = model.generate(**ids, max_length=64, num_beams=4)
    nllb_result = tok.decode(out[0], skip_special_tokens=True)

    # mBART
    ids2 = mbart_tok(inp, return_tensors="pt", truncation=True, max_length=64)
    with torch.no_grad():
        out2 = mbart_model.generate(**ids2, max_length=64, num_beams=4)
    mbart_result = mbart_tok.decode(out2[0], skip_special_tokens=True)

    return nllb_result, mbart_result

# টেস্ট
text = "তুমি কেমন আছো?"
n, m = translate_both(text, direction="dialect", dialect="sylheti")
print(f"\nইনপুট: {text}")
print(f"NLLB : {n}")
print(f"mBART: {m}")
```

```python
import openpyxl
from sentence_transformers import SentenceTransformer, util
import torch

xlsx_path = r"C:\Users\SAMSUNG GALAXY\Desktop\dialect_project\shobuz_cse445.xlsx"

# dataset পড়ি
print("dataset পড়ছি...")
wb = openpyxl.load_workbook(xlsx_path, read_only=True)
ws = wb["Dialect_Dataset"]
rows = list(ws.iter_rows(values_only=True))
header = [str(h).strip() for h in rows[0]]
col = {name: i for i, name in enumerate(header)}

pairs = []  # (standard, dialect, dialect_type)
for r in rows[1:]:
    if r is None or len(r) < 3: continue
    cat = str(r[col.get("Language_Category", 0)] or "").strip()
    sb = str(r[col.get("Standard_Bangla", 1)] or "").strip()
    dl = str(r[col.get("Local_Dialect", 2)] or "").strip()
    if sb and dl:
        pairs.append({"standard": sb, "dialect": dl, "type": cat})

print(f"✅ {len(pairs)} টা জোড়া পড়া হয়েছে")

# embedding মডেল লোড (বাংলা বোঝে)
print("\nembedding মডেল লোড হচ্ছে... (প্রথমবার নামাবে, একটু সময়)")
embedder = SentenceTransformer("paraphrase-multilingual-MiniLM-L12-v2")
print("✅ embedder রেডি")

# সব standard বাক্যের embedding বানাই (RAG index)
print("\nindex বানাচ্ছি...")
standard_texts = [p["standard"] for p in pairs]
dialect_texts = [p["dialect"] for p in pairs]
std_embeddings = embedder.encode(standard_texts, convert_to_tensor=True, show_progress_bar=True)
dl_embeddings = embedder.encode(dialect_texts, convert_to_tensor=True, show_progress_bar=True)
print("✅ RAG index তৈরি!")
```

```python
def rag_correct(input_text, model_output, direction="dialect", threshold=0.75):
    """
    মডেলের অনুবাদ (model_output) কে RAG দিয়ে যাচাই/সংশোধন করে।
    input_text = মূল বাক্য (standard বা dialect)
    direction = "dialect" (std→dialect) বা "standard" (dialect→std)
    """
    # কোন দিকে খুঁজব ঠিক করি
    if direction == "dialect":
        # ইনপুট standard, তাই standard embedding এ খুঁজি
        query_emb = embedder.encode(input_text, convert_to_tensor=True)
        hits = util.cos_sim(query_emb, std_embeddings)[0]
        best_idx = int(hits.argmax())
        best_score = float(hits[best_idx])
        reference = pairs[best_idx]["dialect"]   # সঠিক dialect রূপ
        matched_input = pairs[best_idx]["standard"]
    else:
        # ইনপুট dialect, তাই dialect embedding এ খুঁজি
        query_emb = embedder.encode(input_text, convert_to_tensor=True)
        hits = util.cos_sim(query_emb, dl_embeddings)[0]
        best_idx = int(hits.argmax())
        best_score = float(hits[best_idx])
        reference = pairs[best_idx]["standard"]  # সঠিক standard রূপ
        matched_input = pairs[best_idx]["dialect"]

    # সিদ্ধান্ত: যদি খুব কাছের মিল থাকে (dataset এ প্রায় একই বাক্য),
    # তাহলে dataset এর সঠিক রূপ ব্যবহার করো
    if best_score >= threshold:
        corrected = reference
        used_rag = True
    else:
        corrected = model_output  # মিল কম, মডেলের আউটপুটই রাখো
        used_rag = False

    return {
        "model_output": model_output,
        "corrected": corrected,
        "used_rag": used_rag,
        "match_score": round(best_score, 3),
        "matched_reference": reference,
    }


# টেস্ট করি
text = "তুমি কেমন আছো?"
nllb_out = translate(text, direction="dialect", dialect="sylheti")  # আগের NLLB ফাংশন
result = rag_correct(text, nllb_out, direction="dialect")

print(f"ইনপুট        : {text}")
print(f"মডেল অনুবাদ  : {result['model_output']}")
print(f"RAG সংশোধন   : {result['corrected']}")
print(f"RAG ব্যবহার   : {result['used_rag']}  (মিল স্কোর: {result['match_score']})")
```

```python
def smart_translate(text, direction="dialect", dialect="sylheti", model_choice="nllb"):
    """
    মূল অনুবাদ ফাংশন — মডেল অনুবাদ + RAG সংশোধন একসাথে।
    direction: "dialect" (std→dialect) বা "standard" (dialect→std)
    dialect: sylheti / chittagonian / mymensingh / noakhali
    model_choice: "nllb" বা "mbart"
    """
    # ইনপুট ফরম্যাট
    if direction == "dialect":
        inp = f"<{dialect}> translate to dialect: {text}"
    else:
        inp = f"<{dialect}> translate to standard: {text}"

    # মডেল দিয়ে অনুবাদ
    if model_choice == "nllb":
        ids = tok(inp, return_tensors="pt", truncation=True, max_length=64)
        with torch.no_grad():
            out = model.generate(**ids, max_length=64, num_beams=4)
        model_out = tok.decode(out[0], skip_special_tokens=True)
    else:  # mbart
        ids = mbart_tok(inp, return_tensors="pt", truncation=True, max_length=64)
        with torch.no_grad():
            out = mbart_model.generate(**ids, max_length=64, num_beams=4)
        model_out = mbart_tok.decode(out[0], skip_special_tokens=True)

    # RAG সংশোধন
    rag = rag_correct(text, model_out, direction=direction)

    return {
        "input": text,
        "direction": direction,
        "dialect": dialect,
        "model": model_choice,
        "model_output": model_out,
        "final": rag["corrected"],
        "rag_used": rag["used_rag"],
        "score": rag["match_score"],
    }


# কয়েকটা টেস্ট
tests = [
    ("তুমি কেমন আছো?", "dialect", "sylheti", "nllb"),
    ("আমি ভাত খাবো।", "dialect", "chittagonian", "nllb"),
    ("তুঁই কেমন আছো?", "standard", "sylheti", "nllb"),
]

for text, d, dl, mc in tests:
    r = smart_translate(text, direction=d, dialect=dl, model_choice=mc)
    print(f"\nইনপুট : {r['input']}  [{r['dialect']}, {mc}]")
    print(f"মডেল  : {r['model_output']}")
    print(f"চূড়ান্ত: {r['final']}  (RAG: {r['rag_used']}, স্কোর: {r['score']})")
```

```python
import gradio as gr

DIALECTS = ["sylheti", "chittagonian", "mymensingh", "noakhali"]

def web_translate(text, direction_label, dialect, model_choice):
    if not text.strip():
        return "", "", ""
    # direction ঠিক করি
    direction = "dialect" if direction_label == "Standard → Dialect" else "standard"
    r = smart_translate(text, direction=direction, dialect=dialect, model_choice=model_choice)
    rag_info = f"RAG সংশোধন: {'হ্যাঁ' if r['rag_used'] else 'না'} (মিল স্কোর: {r['score']})"
    return r["model_output"], r["final"], rag_info

with gr.Blocks(title="বাংলা উপভাষা অনুবাদক") as demo:
    gr.Markdown("# 🗣️ বাংলা উপভাষা অনুবাদক\nNLLB / mBART + RAG সংশোধন")

    with gr.Row():
        with gr.Column():
            text_in = gr.Textbox(label="বাক্য লিখুন", placeholder="যেমন: তুমি কেমন আছো?", lines=2)
            direction = gr.Radio(
                ["Standard → Dialect", "Dialect → Standard"],
                value="Standard → Dialect", label="দিক"
            )
            dialect = gr.Dropdown(DIALECTS, value="sylheti", label="উপভাষা")
            model_choice = gr.Radio(["nllb", "mbart"], value="nllb", label="মডেল")
            btn = gr.Button("অনুবাদ করুন", variant="primary")
        with gr.Column():
            model_out = gr.Textbox(label="মডেলের অনুবাদ")
            final_out = gr.Textbox(label="✅ চূড়ান্ত (RAG সংশোধিত)")
            rag_out = gr.Textbox(label="তথ্য")

    btn.click(web_translate,
              inputs=[text_in, direction, dialect, model_choice],
              outputs=[model_out, final_out, rag_out])

    gr.Examples(
        examples=[
            ["তুমি কেমন আছো?", "Standard → Dialect", "sylheti", "nllb"],
            ["আমি ভাত খাবো।", "Standard → Dialect", "chittagonian", "nllb"],
            ["তুঁই কেমন আছো?", "Dialect → Standard", "sylheti", "nllb"],
        ],
        inputs=[text_in, direction, dialect, model_choice],
    )

demo.launch()
```
